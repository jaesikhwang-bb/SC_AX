# MCP 호출

```python
import requests

GENOS_URL = "https://genos.genon.ai"
API_ENDPOINT = f"{GENOS_URL}/api/gateway"

mcp_id = 475
mcp_token = "44fa3a0ee9b34d35a1daed01a2a2944d"
mcp_url = f"{API_ENDPOINT}/mcp/{mcp_id}/mcp"
mcp_name = "test_tool"
```

```python
payload = {
    "jsonrpc" : "2.0",
    "id":"1474",
    "method" : "tools/call",
    "params" : {
        "name":mcp_name,
        "arguments" :{
            "param1" : "data1",
            "param2" : "data2"
        }
    }
}
```

```python
res = requests.post(
    mcp_url,
    headers={
        "Authorization" : f"Bearer {mcp_token}",
        "Accept" : "application/json, text/event-stream",
        "Content-Type" : "application/json"
    },
    json=payload
)
```

```python
# data: 이후만 추출
data = res.text.split("data:", 1)[1].strip()
obj = json.loads(data)
result = obj["result"]["structuredContent"]
print(result)
```
